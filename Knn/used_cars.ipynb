```python
import pandas as pd
import pandas as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("used_car_dataset.csv")
```

```python
df.isnull().sum()
```

```python
df["kmDriven"]=df["kmDriven"].str.replace("km","").str.replace(",","")
```

```python
df["kmDriven"]=df["kmDriven"].astype("float")
```

```python
df["kmDriven"]=df["kmDriven"].fillna(df["kmDriven"].median())
```

```python
df.drop(columns=["Brand","PostedDate","AdditionInfo"],inplace=True)
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
df["FuelType"]=le.fit_transform(df["FuelType"])
```

```python
df["Owner"]=le.fit_transform(df["Owner"])
```

```python
df["model"]=le.fit_transform(df["model"])
```

```python
df["Transmission"]=le.fit_transform(df["Transmission"])
```

```python
df["AskPrice"]=df["AskPrice"].str.replace("₹","")
df["AskPrice"]=df["AskPrice"].str.replace(",","")
df["AskPrice"]=df["AskPrice"].astype("float")
```

```python
df.dtypes
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.30,random_state=10)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsRegressor
```

```python
model=KNeighborsRegressor(n_neighbors=5)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae=mean_absolute_error(y_test,y_pred)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test,y_pred)
```
