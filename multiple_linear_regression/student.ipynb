```python
df=df.drop_duplicates()
```

```python
x=df.drop(columns="G3")
y=df["G3"]
```

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import OneHotEncoder
```

```python
col=make_column_transformer((OneHotEncoder(handle_unknown="ignore"),["school","sex","address","famsize","Pstatus","Mjob","Fjob","reason","guardian","schoolsup","famsup","paid","activities","nursery","higher","internet","romantic"]),remainder="passthrough")
```

```python
x=col.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.20,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score,root_mean_squared_error
```

```python
print(mean_squared_error(y_test,y_pred))
```

```python
print(mean_absolute_error(y_test,y_pred))
```

```python
print(r2_score(y_test,y_pred))
```

```python
print(root_mean_squared_error(y_test,y_pred))
```
