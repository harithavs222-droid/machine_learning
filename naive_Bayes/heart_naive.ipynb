```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("heart__missing.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df.isna().sum()
```

```python
cate=["age","sex","cp","trestbps","chol","fbs","restecg","thalach","exang","oldpeak","slope","ca","thal"]
```

```python
for  i  in cate:
    df[i]=df[i].fillna(df[i].mean())
```

```python
df.isna().sum()
```

```python
df.columns
```

```python
df=df.drop_duplicates()
```

```python
x=df.drop(columns="target")
y=df["target"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.25,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score

knn=KNeighborsClassifier(n_neighbors=5)
svm=SVC()
naive=GaussianNB()
```

```python
lst=[knn,svm,naive]

for i in lst:
    print("model selection______________",i)
    print(i.fit(x_train,y_train))
    y_pred=i.predict(x_test)

    print("confusion matrix______________",i)
    print(confusion_matrix(y_pred,y_test))

    print("accuracy score________________",i)
    print(accuracy_score(y_pred,y_test))
```
