```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("poly_regression_data.csv")
df
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
```

```python
plt.scatter(x=x,y=y)
```

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

```python
model.fit(x,y)
```

```python
y_pred=model.predict(x)
y_pred
```

```python
y
```

```python
from sklearn.metrics import mean_absolute_error,r2_score,mean_absolute_percentage_error
```

```python
print(mean_absolute_percentage_error(y_pred,y))
```

```python
print(mean_absolute_error(y_pred,y))
```

```python
print(r2_score(y_pred,y))
```

```python
plt.scatter(x=x,y=y)
plt.plot(x,y_pred)
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly=PolynomialFeatures(degree=2)
```

```python
x_poly=poly.fit_transform(x)
x_poly
```

```python
x_poly.shape
```

```python
model1=LinearRegression()
model1.fit(x_poly,y)
```

```python
y_pred1=model1.predict(x_poly)
```

```python
print(mean_absolute_error(y,y_pred1))
```

```python
print(mean_absolute_percentage_error(y,y_pred1))
```

```python
print(r2_score(y,y_pred1))
```

```python
plt.scatter(x,y,color="m")
plt.plot(x,y_pred1,color="g")
```
