```python
df.drop(columns="Loan_ID",inplace=True)
```

```python
cate=["Gender","Married","Dependents","Self_Employed","Loan_Amount_Term","Credit_History"]
```

```python
for i in cate:
    df[i]=df[i].fillna(df[i].mode()[0])
```

```python
df["LoanAmount"]=df["LoanAmount"].fillna(df["LoanAmount"].mean())
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
columns=["Gender","Married","Dependents","Education","Self_Employed","Property_Area","Loan_Status"]
```

```python
for col in columns:
    df[col]=le.fit_transform(df[col])
```

```python
df.dtypes
```

```python
x=df.drop(columns="Loan_Status")
y=df["Loan_Status"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.30,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix
from sklearn.metrics import  accuracy_score
knn=KNeighborsClassifier(n_neighbors=5)
svm=SVC()

lst=[knn,svm]
```

```python
for i in lst:
    print("model selection________________",i)
    print(i.fit(x_test,y_test))
    y_pred=i.predict(x_test) 



    print("confusion_matrix______________")
    print(confusion_matrix(y_pred,y_test))


    print("accuracy_score_________________")
    print(accuracy_score(y_pred,y_test))
```
